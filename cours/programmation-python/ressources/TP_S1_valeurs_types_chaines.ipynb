## 1. Variables et types

Une **variable** est un nom auquel on attache une valeur, avec le signe `=`. Ce signe n'est pas l'égalité des mathématiques : c'est une **affectation** (« range cette valeur sous ce nom »). Chaque valeur a un **type** : texte (`str`), entier (`int`), décimal (`float`) ou booléen (`bool`).

```python
nom = "Fatou"
age = 23
moyenne = 14.5
ville = "Dakar"
est_etudiant = True

print(nom)
print(age)
print(moyenne)
print(ville)
print(est_etudiant)
```

La fonction `type()` révèle le type d'une valeur. Python le devine tout seul à partir de la façon dont la valeur est écrite (guillemets, point décimal, `True`/`False`).

```python
print(type(nom))
print(type(age))
print(type(moyenne))
print(type(est_etudiant))
```

## 2. Opérations

On calcule avec les opérateurs habituels. Deux subtilités utiles : `/` donne **toujours** un résultat décimal, `//` est la **division entière** (on garde la partie entière), et `%` donne le **reste**.

```python
a = 7
b = 2
print("somme        :", a + b)
print("difference   :", a - b)
print("produit      :", a * b)
print("division     :", a / b)
print("div. entiere :", a // b)
print("reste        :", a % b)
print("puissance    :", a ** b)
```

Les **comparaisons** (`==`, `!=`, `<`, `>`, `<=`, `>=`) produisent un booléen : `True` ou `False`. Attention : `=` affecte, `==` compare.

```python
x = 14
print(x >= 10)
print(x == 10)
print(x != 10)
```

On combine des conditions avec `and` (et), `or` (ou), `not` (non).

```python
note = 12
assiduite = True
admis = note >= 10 and assiduite
print("Admis ?", admis)
print("Pas admis ?", not admis)
```

## 3. Afficher proprement avec les f-strings

Une **f-string** (préfixe `f` devant les guillemets) insère des variables dans du texte, entre accolades `{ }`. C'est la façon propre d'afficher un message qui mélange texte et valeurs.

```python
prenom = "Awa"
ville = "Thiès"
print(f"Je m'appelle {prenom} et j'habite à {ville}.")
```

On peut aussi **mettre en forme** une valeur dans l'accolade. Par exemple `:.2f` affiche un décimal avec deux chiffres après la virgule.

```python
moyenne = 14.567
print(f"Moyenne : {moyenne:.2f}")
```

Rappel de la Séance 0 : l'opérateur `+` ne « colle » que du **texte**. Coller un nombre à du texte avec `+` provoque une `TypeError`. La f-string règle le problème en insérant la valeur entre accolades, comme dans `f"Age : {age}"`.

## 4. Les chaînes de caractères

Une chaîne (`str`) est une **séquence de caractères**. `len()` donne sa longueur. On accède à un caractère par sa **position** entre crochets : la position commence à **0**, et `-1` désigne le dernier.

```python
ville = "Saint-Louis"
print("longueur :", len(ville))
print("premier  :", ville[0])
print("dernier  :", ville[-1])
```

On extrait un morceau par **découpage** (`slicing`) : `[debut:fin]` prend les caractères de `debut` jusqu'à `fin` **exclu**. On peut omettre l'un des deux bornes.

```python
ville = "Saint-Louis"
print(ville[0:5])
print(ville[6:])
print(ville[:5])
```

Les chaînes offrent des **méthodes** (des outils attachés à la valeur, appelés avec un point). Les plus utiles : `.strip()` enlève les espaces aux bords, `.upper()` / `.lower()` changent la casse, `.replace(a, b)` remplace.

```python
phrase = "  Bonjour Dakar  "
print(phrase.strip())
print(phrase.strip().upper())
print(phrase.strip().lower())
print(phrase.strip().replace("Dakar", "Thiès"))
```

Important : une méthode de chaîne renvoie une **nouvelle** chaîne ; l'originale ne change pas (on dit qu'une chaîne est **immuable**).

```python
mot = "dakar"
print(mot.upper())
print(mot)
```

Enfin, `.split(sep)` découpe une chaîne en une **liste** de morceaux, selon un séparateur. C'est très utilisé pour lire des données.

```python
villes = "Dakar,Thiès,Kaolack".split(",")
print(villes)
print("nombre de villes :", len(villes))
```

## Récapitulatif

- Une **variable** (`=`) attache une valeur à un nom ; chaque valeur a un **type** (`str`, `int`, `float`, `bool`), révélé par `type()`.
- Les **opérations** : arithmétique (`+ - * / // % **`), comparaisons et booléens (`and`/`or`/`not`) produisent `True`/`False`.
- Les **f-strings** (`f"... {variable} ..."`) affichent proprement texte et valeurs ; `:.2f` met en forme.
- Une **chaîne** est une séquence : `len`, indexation `[i]`, découpage `[a:b]`, méthodes `.strip/.upper/.lower/.replace/.split` ; elle est **immuable**.
