## 1. Les fonctions

Une **fonction** est un bloc de code **nommé** et **réutilisable**. On la définit une fois avec `def`, puis on l'**appelle** autant de fois qu'on veut. `return` renvoie une valeur que l'on peut réutiliser. Comme pour `if` et `for`, le corps est un **bloc indenté** après les deux points.

```python
def moyenne(a, b):
    return (a + b) / 2

print(moyenne(12, 16))
print(moyenne(9, 18))
```

Une fonction peut prendre **plusieurs paramètres**. À l'appel, on fournit les **arguments** dans le même ordre.

```python
def message(prenom, ville):
    return f"{prenom} habite a {ville}."

print(message("Fatou", "Dakar"))
```

Un paramètre peut avoir une **valeur par défaut** : si l'argument n'est pas fourni à l'appel, la valeur par défaut est utilisée.

```python
def saluer(nom, politesse="Bonjour"):
    return f"{politesse} {nom}"

print(saluer("Awa"))
print(saluer("Awa", "Bonsoir"))
```

On **documente** une fonction avec une **docstring** : une courte phrase entre guillemets, juste sous le `def`. On la retrouve via `.__doc__`.

```python
def carre(x):
    "Renvoie le carre de x."
    return x * x

print(carre(5))
print(carre.__doc__)
```

Une variable créée **dans** une fonction est **locale** : elle n'existe que le temps de l'exécution de la fonction, et n'est pas accessible à l'extérieur (c'est la **portée**).

```python
def calcul():
    resultat = 5 * 3
    print("dans la fonction :", resultat)

calcul()
```

> **Portée.** La variable `resultat` ci-dessus n'existe **pas** en dehors de `calcul` : y accéder après l'appel provoquerait une `NameError`. Pour récupérer une valeur calculée dans une fonction, il faut la **renvoyer** avec `return`.

Attention à la différence entre **`return`** et **`print`** : `return` **renvoie** une valeur réutilisable dans un calcul ; une fonction qui se contente d'afficher renvoie `None`.

```python
def somme(a, b):
    return a + b

def affiche_somme(a, b):
    print(a + b)

x = somme(10, 5) + 100
print("x =", x)
y = affiche_somme(10, 5)
print("y =", y)
```

## 2. Les modules

Un **module** est une bibliothèque de code déjà écrit, qu'on **importe** pour la réutiliser. Python en fournit beaucoup en standard. `import math` donne accès aux outils mathématiques via `math.`.

```python
import math

print(math.sqrt(16))
print(math.pi)
```

**À vous.** Importez `math` et affichez la racine carrée de 81 avec `math.sqrt`.

```python
# À vous : importez math et affichez math.sqrt(81)
```

On peut aussi importer **seulement** ce dont on a besoin, avec `from module import nom`. On l'utilise alors directement, sans le préfixe.

```python
from math import sqrt

print(sqrt(25))
```

## 3. Lire et écrire des fichiers

Un programme peut **sauvegarder** des données dans un fichier pour les retrouver plus tard. On ouvre un fichier avec `open()`, dans un bloc `with` qui le **referme automatiquement** à la fin. Le mode `"w"` **écrit** (et écrase), `"r"` **lit** (mode par défaut).

```python
with open("villes.txt", "w") as f:
    f.write("Dakar\n")
    f.write("Louga\n")
    f.write("Mbour\n")

print("fichier ecrit")
```

`\n` marque une fin de ligne. On relit tout le contenu d'un coup avec `.read()`.

```python
with open("villes.txt") as f:
    contenu = f.read()

print(contenu)
```

Le plus souvent, on parcourt un fichier **ligne par ligne** avec une boucle `for`. Chaque ligne se termine par `\n`, qu'on enlève avec `.strip()`.

```python
with open("villes.txt") as f:
    for ligne in f:
        print(ligne.strip())
```

## 4. Gérer les erreurs : try / except

Certaines opérations peuvent **échouer** à l'exécution (convertir un texte non numérique, diviser par zéro, ouvrir un fichier absent). Plutôt que de laisser le programme planter, on **essaie** le code dans un bloc `try`, et on **rattrape** l'erreur dans un bloc `except`.

```python
try:
    nombre = int("abc")
    print(nombre)
except ValueError:
    print("Ce n'est pas un nombre valide.")
```

**À vous.** Mettez `int("12x")` dans un `try`, et affichez un message dans un `except ValueError`.

```python
# À vous : try: int("12x") ... except ValueError: affichez un message
```

On précise **quelle** erreur on rattrape (ici `ZeroDivisionError`). Cela évite de masquer des erreurs qu'on n'avait pas prévues.

```python
try:
    resultat = 10 / 0
except ZeroDivisionError:
    print("Division par zero impossible.")
```

Une fonction peut intégrer un `try / except` pour rester **robuste** : elle renvoie une valeur de repli au lieu de planter.

```python
def division_sure(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "impossible"

print(division_sure(10, 2))
print(division_sure(10, 0))
```

## Récapitulatif

- **Fonction** : `def nom(parametres):` + `return` ; s'appelle avec des **arguments** ; peut avoir des **valeurs par défaut** et une **docstring**.
- **Portée** : une variable définie dans une fonction est **locale** ; pour en sortir une valeur, on la `return`.
- **return vs print** : `return` renvoie une valeur réutilisable ; `print` affiche seulement (et la fonction renvoie `None`).
- **Modules** : `import math` (puis `math.sqrt`), ou `from math import sqrt`.
- **Fichiers** : `with open("f.txt", "w") as f:` pour écrire, `"r"` pour lire ; `.read()` ou une boucle `for` ligne par ligne ; `with` referme automatiquement.
- **Exceptions** : `try:` le code risqué, `except TypeErreur:` le rattrapage ; préciser l'erreur attendue.
