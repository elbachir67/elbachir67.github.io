## 1. Les conditions : if / else

Une condition exécute un bloc de code **seulement si** un test est vrai. Le test est un booléen (rappel de la Séance 1). La syntaxe repose sur deux points `:` et sur l'**indentation** (le décalage) : le bloc décalé appartient au `if`.

```python
note = 12
if note >= 10:
    print("Admis")
else:
    print("Recale")
```

**À vous.** Avec `temperature = 34`, affichez `"Chaud"` si la température dépasse 30, sinon `"Doux"`.

```python
# À vous : si temperature > 30 afficher "Chaud", sinon "Doux"
```

Pour distinguer **plusieurs cas**, on enchaîne avec `elif` (sinon si). Python teste les conditions dans l'ordre et s'arrête à la première vraie.

```python
note = 14
if note >= 16:
    mention = "Tres bien"
elif note >= 14:
    mention = "Bien"
elif note >= 10:
    mention = "Passable"
else:
    mention = "Insuffisant"
print(mention)
```

**À vous.** Avec `age = 25`, affichez `"Enfant"` si l'âge est inférieur à 18, `"Adulte"` s'il est inférieur à 65, sinon `"Senior"`.

```python
# À vous : if age < 18 "Enfant", elif age < 65 "Adulte", else "Senior"
```

On peut combiner des conditions avec `and`, `or`, `not` (Séance 1). Le bloc n'est indenté qu'une fois — l'oubli des deux points `:` ou de l'indentation est l'erreur la plus fréquente ici.

```python
note = 15
assiduite = True
if note >= 10 and assiduite:
    print("Valide")
else:
    print("A revoir")
```

## 2. La boucle for

Une boucle `for` répète un bloc **pour chaque élément** d'une collection. À chaque tour, la variable de boucle prend la valeur de l'élément courant.

```python
villes = ["Dakar", "Louga", "Mbour"]
for ville in villes:
    print(ville)
```

Pour répéter un nombre fixé de fois, on utilise `range()`. `range(n)` produit les entiers de `0` à `n-1` ; `range(a, b)` va de `a` à `b` exclu.

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 4):
    print("tour", i)
```

Le motif le plus utile est l'**accumulation** : on prépare une variable *avant* la boucle, puis on la met à jour à chaque tour. Ici, on calcule une somme.

```python
notes = [12, 15, 9, 18]
total = 0
for note in notes:
    total = total + note
print("total :", total)
print("moyenne :", total / len(notes))
```

Même principe pour **compter** : on incrémente un compteur quand une condition est remplie.

```python
villes = ["Dakar", "Saint-Louis", "Mbour", "Ziguinchor"]
grandes = 0
for ville in villes:
    if len(ville) > 5:
        grandes = grandes + 1
print("villes de plus de 5 lettres :", grandes)
```

## 3. Parcourir intelligemment : enumerate et zip

`enumerate()` donne à la fois l'**indice** et la **valeur** de chaque élément — pratique pour numéroter.

```python
villes = ["Dakar", "Louga", "Mbour"]
for i, ville in enumerate(villes):
    print(i, ville)
```

**À vous.** Affichez chaque prénom de `prenoms = ["Fatou", "Awa", "Cheikh"]` précédé de son indice, avec `enumerate`.

`zip()` parcourt **deux listes en parallèle**, élément par élément — idéal pour associer des données qui se correspondent.

```python
villes = ["Dakar", "Thies", "Kaolack"]
population = [1200000, 320000, 240000]
for ville, hab in zip(villes, population):
    print(ville, ":", hab)
```

## 4. La boucle while

Une boucle `while` répète un bloc **tant qu'**une condition reste vraie. Il faut absolument faire évoluer cette condition à l'intérieur de la boucle, sinon elle ne s'arrête jamais.

```python
compteur = 0
while compteur < 3:
    print("compteur =", compteur)
    compteur = compteur + 1
print("fin")
```

> **Piège — la boucle infinie.** Si la condition ne devient jamais fausse (par exemple si on oublie `compteur = compteur + 1`), la boucle tourne sans fin. Vérifiez toujours qu'une variable avance **vers** la sortie. On interrompt une boucle emballée avec le bouton *Interrupt* du noyau.

## 5. Contrôler une boucle : break et continue

`break` **arrête** complètement la boucle ; `continue` **saute** au tour suivant sans exécuter la suite du bloc.

```python
for n in [1, 2, 3, 4, 5]:
    if n == 3:
        break
    print(n)
```

Ici la boucle s'arrête dès que `n` vaut 3. Avec `continue`, au contraire, on saute certains éléments mais on continue :

```python
for n in [1, 2, 3, 4, 5]:
    if n % 2 == 0:
        continue
    print(n)
```

## 6. Retour sur les compréhensions

Une **compréhension de liste** (Séance 2) est simplement une boucle `for` écrite de façon compacte. Les deux versions ci-dessous produisent exactement le même résultat.

```python
# Version longue : boucle for + append
carres = []
for n in [1, 2, 3, 4]:
    carres.append(n * n)
print(carres)

# Version compacte : comprehension
carres2 = [n * n for n in [1, 2, 3, 4]]
print(carres2)
```
